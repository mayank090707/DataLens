DataLens – Student Performance Analysis and Data Visualization

Abstract:
DataLens is a Python-based data analysis and visualization project developed to explore and understand student academic performance using the Student Performance dataset.

The project focuses on Exploratory Data Analysis (EDA), data cleaning, statistical analysis and data visualization to identify patterns and relationships among different student-related factors.

The dataset contains information about students, including age, gender, study time, school absences, previous academic failures and examination grades.

Various Python libraries, including Pandas, NumPy, Matplotlib and Seaborn, are used to process, analyze and visualize the data. Five major visualizations are developed to investigate grade distributions, study habits, school absences, numerical correlations, and performance differences across gender and schools.

An interactive dashboard is also developed using Streamlit. It allows users to apply filters, explore student records, view calculated performance indicators and download filtered datasets.

The primary objective of DataLens is to transform raw student data into meaningful information through statistical analysis and interactive visualization, making academic performance patterns easier to understand.

1. Problem Statement
Student academic performance is influenced by several factors, including study habits, school attendance, previous academic performance and personal circumstances.
Understanding the relationship between these factors can be challenging when dealing with large amounts of raw student data.

Without proper analysis and visualization, identifying patterns, comparing student performance and extracting meaningful insights becomes difficult.

Therefore, this project aims to analyze a Student Performance dataset using Python-based data analysis and visualization techniques.

The project focuses on cleaning raw data, performing exploratory data analysis, calculating statistical measures and creating visualizations to identify relationships between different student-related factors.

An interactive dashboard is also developed to make the analysis accessible and easier to understand.

6. Project Methodology and Workflow

The DataLens project follows a systematic approach to analyze student performance and present meaningful insights through data visualization.
Step 1: Dataset Collection
The Student Performance dataset was collected from the UCI Machine Learning Repository. The mathematics dataset (`student-mat.csv`) was selected for analysis.
Step 2: Data Loading
The dataset was imported into the Jupyter Notebook using the Pandas library. The data was loaded into a Pandas DataFrame for further processing and analysis.
Step 3: Data Understanding
Initial exploration was performed to understand the structure and characteristics of the dataset. This included:
- Displaying the first few records.
- Checking column names and data types.
- Examining statistical summaries.
- Identifying missing values and duplicate records.
Step 4: Data Cleaning and Preprocessing
The dataset was checked for missing values and duplicate records. A separate cleaned copy of the dataset was created to maintain the original data. The cleaned dataset was then saved as a CSV file for further use.
Step 5: Exploratory Data Analysis (EDA)
Exploratory Data Analysis was performed to identify patterns and relationships in student performance. Statistical measures such as mean, median, mode, minimum, maximum, and standard deviation were calculated.
Step 6: Data Visualization
Five visualizations were created using Matplotlib and Seaborn:
- Histogram to understand the distribution of final grades.
- Box plot to study the relationship between study time and final grades.
- Scatter plot to examine the relationship between absences and final grades.
- Correlation heatmap to identify relationships among numerical variables.
- Grouped bar chart to compare average grades across schools and genders.
Step 7: Performance Analysis
Students were categorized into performance groups based on their final mathematics grade (G3). Pass percentage, average grade, and the number of students in different performance categories were calculated.
Step 8: Interactive Dashboard Development
A web-based interactive dashboard was developed using Streamlit. It provides:
- Key performance indicators.
- Interactive filters for school, gender, study time, and age.
- Data visualizations based on filtered records.
- A student data explorer.
- An option to download filtered data as a CSV file.
Step 9: Testing and Validation
The dashboard was tested by changing different filter combinations and checking whether the displayed statistics and visualizations updated correctly.

Project Workflow
Dataset Collection → Data Loading → Data Understanding → Data Cleaning → Exploratory Data Analysis → Data Visualization → Performance Analysis → Streamlit Dashboard → Testing and Validation



7. Data Analysis and Implementation
This section describes the implementation of data analysis techniques using Python libraries. The dataset was processed using Pandas and NumPy, while Matplotlib and Seaborn were used to visualize the results.
7.1 Dataset Loading and Inspection

- Standard Deviation: Measures how much the grades vary from the mean.

Visualization Techniques

```python

# Select only numerical columns
numeric_df = df.select_dtypes(include="number")
# Calculate correlation matrix
correlation_matrix = numeric_df.corr()
# Create heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap of Student Performance Dataset")
plt.tight_layout()
plt.show()
```

```python

# Find correlation of all numerical features with G3
grade_correlation = correlation_matrix["G3"].sort_values(
    ascending=False
)
print("Correlation with Final Grade (G3):")
print(grade_correlation.round(3))
```

```python

# Save cleaned dataset
clean_df.to_csv("../data/cleaned_students.csv", index=False)
print("Cleaned dataset saved successfully!")
```

```python

# Statistical analysis of final grades
print("FINAL GRADE STATISTICS")
print("Mean:", round(clean_df["G3"].mean(), 2))
print("Median:", clean_df["G3"].median())
print("Mode:", clean_df["G3"].mode().tolist())
print("Minimum Grade:", clean_df["G3"].min())
print("Maximum Grade:", clean_df["G3"].max())
print("Standard Deviation:",
      round(clean_df["G3"].std(), 2))
```

```python

# Categorize students based on final grades
def categorize_grade(grade):
    if grade >= 16:
        return "Excellent"
    elif grade >= 12:
        return "Good"
    elif grade >= 10:
        return "Average"
    else:
        return "Below Pass"
clean_df["Performance"] = clean_df["G3"].apply(
    categorize_grade
)
# Count students in each category
performance_counts = clean_df["Performance"].value_counts()
print("Student Performance Categories:")
print(performance_counts)
```

```python

plt.figure(figsize=(9, 5))
sns.countplot(
    data=clean_df,
    x="Performance",
    order=["Below Pass", "Average", "Good", "Excellent"],
    hue="Performance",
    palette="Set2",
    legend=False
)
plt.title("Distribution of Student Performance Categories")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

total_students = len(clean_df)
average_grade = clean_df["G3"].mean()
passed_students = (clean_df["G3"] >= 10).sum()
failed_students = (clean_df["G3"] < 10).sum()
pass_percentage = (passed_students / total_students) * 100
print("OVERALL STUDENT PERFORMANCE")
print("--------------------------------")
print("Total Students:", total_students)
print("Average Grade:", round(average_grade, 2))
print("Passed Students:", passed_students)
print("Below-Pass Students:", failed_students)
print("Pass Percentage:", round(pass_percentage, 2), "%")
```
